## Ans 6. Median value of horsepower

```python
# Find the median value of the horsepower column in the dataset.
print("Median horsepower: ", data['horsepower'].median())

# Next, calculate the most frequent value of the same horsepower column.
print("Most frequent horsepower: ", data['horsepower'].mode()[0])

print("No. of NaNs in column:",data['horsepower'].isna().sum())

# Use the fillna method to fill the missing values in the horsepower column with the most frequent value from the previous step.
data['horsepower']  = data['horsepower'].fillna(data['horsepower'].mode()[0])
# print(tmp.head(20))

# Now, calculate the median value of horsepower once again.
# Has it changed?
print("Median horsepower after replacement: ", data['horsepower'].median())
```
