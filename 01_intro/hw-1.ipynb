## Ans 7. Sum of weights

```python
# Select all the cars from Asia
# print(data.columns)
data_asia = data[ data['origin'] == 'Asia' ]
# print(data_asia)

# Select only columns vehicle_weight and model_year
data_asia_subset = data_asia[['vehicle_weight','model_year']]
# print(data_asia_subset)

# Select the first 7 values
data_asia_first_7 = data_asia_subset.head(7)
# print(data_asia_first_7)

# Get the underlying NumPy array. Let's call it X.
X = data_asia_first_7.values
# print(X.dtype)

# Compute matrix-matrix multiplication between the transpose of X and X. To get the transpose, use X.T. Let's call the result XTX.
# Build functions for matrix matrix multiplication

# Vector dot product
def vector_vector_multiplication(v1, v2):
    assert v1.shape == v2.shape, "Vectors must be of the same shape"
    result = 0
    for i in range(v1.shape[0]):
        result += v1[i] * v2[i]

    return result

# Matrix-vector multiplication
def matrix_vector_multiplication(M, v):
    assert M.shape[1] == v.shape[0], "Matrix columns must match vector size"

    result = np.zeros(M.shape[0])
    for i in range(M.shape[0]):
        result[i] = vector_vector_multiplication(M[i, :], v)

    return result

# Matrix-matrix multiplication
def matrix_matrix_multiplication(M1, M2):
    assert M1.shape[1] == M2.shape[0], "Matrix dimensions must match for multiplication"

    result = np.zeros((M1.shape[0], M2.shape[1]))
    for i in range(M1.shape[0]):
        for j in range(M2.shape[1]):
            result[i, j] = vector_vector_multiplication(M1[i, :], M2[:, j])

    return result


XTX = matrix_matrix_multiplication(X.transpose(), X)
print("XTX:\n", XTX)


# Matrix-matrix multiplication using NumPy
XTX = X.T @ X 
print("XTX (NumPy):\n", XTX)

# Invert XTX.
XTX_inv = np.linalg.inv(XTX)
print("Inverse of XTX:\n", XTX_inv)

# Create an array y with values [1100, 1300, 800, 900, 1000, 1100, 1200].
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

# Multiply the inverse of XTX with the transpose of X, and then multiply the result by y. Call the result w.
w = XTX_inv @ X.T @ y

# What's the sum of all the elements of the result?
sum_w = np.sum(w)
print("Sum of elements in w:", sum_w)

# Note: You just implemented linear regression. We'll talk about it in the next lesson.
```
